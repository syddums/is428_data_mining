```python
import pandas as pd # pandas package
movies = pd.DataFrame({ # pd.DataFrame({}) where we create our own dataset via DataFrame
    'title':['Interstellar','Inception','The Matrix','The Martian','Gravity'], # movie titles
    'description': ['Explorers attempt to save humanity by going through wormhole',
                    'Thief steals corporate secrets to erase his past',
                    'Hacker learns reality and fights against computers',
                    'Astronaut lives in Mars',
                    'Astronauts try to survive after fatal accident'], # movie descriptions
})
movies # showcase the DataFrame
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer # TfidVectorizer package
vectorizer = TfidfVectorizer(stop_words='english') # variable that converts english text descriptions into TF-IDF vectors
tfidf_matrix = vectorizer.fit_transform(movies['description']) # matrix that builds the vocabulary table of all english text words
tfidf_matrix # shows tfidf_matrix

# outcome is 5 rows by 39 columns (since each word is a vector) 
```

```python
from sklearn.metrics.pairwise import cosine_similarity # cosine similarity package

similarity_matrix = cosine_similarity(tfidf_matrix,tfidf_matrix) # uhh not sure!

similarity_matrix # shows similarity matrix

# outcome is a binary table
```

```python
movie = 'The Matrix' # variable holds string 'Interstellar'
idx = movies[movies['title']==movie].index[0] # i'm not sure what this does

for i, title in enumerate(movies['title']): # checks for title in the title column in movies dataframe
    print(f"{title:12s} similarity = {similarity_matrix[idx,i]: .4f}") # where for every 12 seconds, check if the the idx 
```

```python
# user-based collaborative filtering
import numpy as np # numpy package
ratings = pd.DataFrame( # ratings DataFrame
    {
    'Interstellar': [5,4,np.nan,5],
    'Inception': [4,5,2,5],
    'The Matrix': [1,2,5,np.nan],
    'The Martian': [5,5,2,4],
    'Gravity': [4,3,np.nan,4]  
    },
    index=['User A','User B','User C','User D'] # rows that contain rating information
)
ratings
```

```python
ratings_filled = ratings.fillna(0) # something to use in the future to fill NaN with system average or 0s
ratings
```

```python
similarity = cosine_similarity(ratings_filled) # apply cosine_similarity on ratings_filled df

similarity_df = pd.DataFrame(similarity, index=ratings.index, columns=ratings.index) # have spaces cuz that.. corrected that
# 

similarity_df # shows table
```

```python
sims=similarity_df['User C'].drop('User C') # idk 
ratings_other=ratings['Interstellar'].drop('User C') # idk
# mask=ratings_other.notna()
result = np.sum(sims*ratings_other)/np.sum(abs(sims)) # calculation for the weighted average
result # show result 
```

```python
def predict_rating(UserName,movieName): # same thing as above but in function
    sims=similarity_df[UserName].drop(UserName) # idk 
    ratings_other=ratings[movieName].drop(UserName) # idk
    result = np.sum(sims*ratings_other)/np.sum(abs(sims)) 
    return result
```

```python
output = predict_rating('User D','Gravity') # calling function with parameters AND CASE SENSITIVE
output # print output
```
